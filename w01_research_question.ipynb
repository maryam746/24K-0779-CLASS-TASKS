## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

I chose this because the starter dataset explicitly tracks trend_direction (growing, declining, stable). This is a high-impact business problem—if we can catch a page before it drops significantly, a content team can rewrite it and save its traffic. It directly connects to saving organic revenue for a client.

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Cost of a wrong call:

False Positive (Flagging a good page): We waste 2 hours of an editor's time rewriting a page that would have stayed stable anyway.

False Negative (Missing a bad page): We lose thousands of clicks and potential sales because we didn't catch the decline in time.
Why ML, not just a rule? A simple rule like "rewrite pages older than 1 year" is too blunt. Some evergreen content ages perfectly; some new articles drop immediately. ML can learn the combination of factors (current position, backlinks, content type, age) that signal a true decline, which a human can't easily track across 30,000 pages.



## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: How big is the problem? (% of pages declining)
decline_pct = df['trend_direction'].value_counts(normalize=True) * 100
print("1. Trend Distribution (%):")
print(decline_pct.round(2))
print("---")

# Number 2: Do declining pages have worse positions? (Median position)
pos_by_trend = df.groupby('trend_direction')['avg_position'].median()
print("2. Median Average Position by Trend:")
print(pos_by_trend.round(1))
print("---")

# Number 3: Is CTR different for declining pages?
ctr_by_trend = df.groupby('trend_direction')['ctr'].median()
print("3. Median CTR by Trend:")
print(ctr_by_trend.round(4))
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

What I can claim: Based on the data above, I can claim a strong association. Pages that are currently 'declining' tend to have a worse average position (e.g., ~15 vs ~8) and a significantly lower Click-Through Rate compared to 'growing' pages.
What I cannot claim: I cannot claim causation. I cannot say "having a high position causes a page to grow" because the relationship goes both ways (good pages get better positions). This is a predictive project, not a causal one. I am simply identifying signals of decline, not proving why Google ranks them lower.
